## Problem 1

Consider the following LP

**Formulation**

The formulation is
$$
\begin{array}{rll}
    \max & \displaystyle x_1 + 3x_2 & \\[15pt]
    \text{s.t.} & -x_1 + x_2 \leq 3 \\[15pt]
    & -x_1 +2x_2 \leq 8\\[15pt]
    & 3x_1 +x_2 \leq 18\\[15pt]
    & x_{i} \geq 0 & \forall i=1,2
\end{array}
$$

```python
import pandas as pd
from pathlib import Path
from itertools import product
#import libraries to work with pyomo
from pyomo.environ import *
from pyomo.opt import SolverFactory
```

**Read Data**

Use pyomo to solve optimization model

```python
m1 = ConcreteModel()

#Decision variables
m1.x1 = Var(domain=NonNegativeReals, name="x1")
m1.x2 = Var(domain=NonNegativeReals, name="x2")

#Constraints
m1.c1 = Constraint(expr= -1*m1.x1 + m1.x2 <= 3)
m1.c2 = Constraint(expr= -1*m1.x1 + 2*m1.x2 <= 8)
m1.c3 = Constraint(expr= 3*m1.x1 + m1.x2 <= 18)

#Objective function
m1.obj = Objective(expr= m1.x1 + 3*m1.x2, sense=maximize)
```

```python
#solve the model
solver = SolverFactory("gurobi")
results = solver.solve(m1, tee=True)
#display results
print("Optimal Result:", value(m1.obj))
print("x1:", value(m1.x1))
print("x2:", value(m1.x2))
```
